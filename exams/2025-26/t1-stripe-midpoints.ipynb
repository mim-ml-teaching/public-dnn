# Exam 2025/26 task 1: Stripe midpoints

You are given a 4D tensor with a batch of images, shape $(B, C, H, W)$. The images have only one channel (C=1), with each pixel being either black or white (0 or 1). Your task is to output a tensor with the same shape, with zeros everywhere except for *midpoints of horizontal segments* appearing in the input image. For these midpoints, the value should be equal to one.

A horizontal stripe is a contiguous horizontal sequence of pixels with values equal to 1.
The for a stripe starting at position $(b, 1, h, w_1)$ and ending at position $(b, 1, h, w_2)$ (inclusive), the coordinates of its midpoint are $(b, 1, h, \lfloor\frac{w_1+w_2}{2}\rfloor)$

**Important!**
* You cannot use loops, parallelized function calling (map/vmap), nor recursion.
* All import statements have been provided for you, do not add anything extra.

You are given:
* A utility function to generate example input images,
* Example unit tests to check if the solution is right.

```python
import torch
import matplotlib.pyplot as plt
```

```python
def generate_images(batch_size: int, image_size: int) -> torch.Tensor:
    canvas = torch.zeros((batch_size, 1, image_size, image_size))

    for i in range(batch_size):
        canvas[i, :, :, :] = generate_image(image_size)

    return canvas


def generate_image(image_size: int) -> torch.Tensor:
    canvas = torch.zeros((1, image_size, image_size))
    num_segments = int(torch.randint(5, 10, (1,)).item())

    for _ in range(num_segments):
        row = int(torch.randint(0, image_size, (1,)).item())
        col = int(torch.randint(0, image_size, (1,)).item())

        length = int(torch.randint(1, image_size - col + 1, (1,)).item())

        canvas[0, row, col : col + length] = 1.0

    return canvas


def visualize_datapoint(x: torch.Tensor) -> None:
    plt.imshow(x.squeeze().numpy(), cmap="gray")
    plt.axis("off")
    plt.show()
```

```python
# Batches are randomly generated.
example_batch = generate_images(batch_size=1, image_size=10)
visualize_datapoint(example_batch[0])
```
